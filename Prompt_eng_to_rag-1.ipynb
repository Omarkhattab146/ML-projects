### LCEL & Runnables Basics

```python
# convert simple function in python into runnable
from langchain_core.runnables import RunnableLambda

def double(x):
    return 2*x

runnable_double = RunnableLambda(double) # the new line
print(runnable_double.invoke(6)) # 6 -> 12
print(runnable_double.batch([2, 4, 6, 8])) # [2, 4, 6, 8] -> [4, 8, 12, 16]

multiply_by_eight = runnable_double | runnable_double | runnable_double
print(multiply_by_eight.invoke(1))  # 1 ->2 ->4 ->8 
```

```python
# LCEL operator is | .
import re
from langchain_core.output_parsers import StrOutputParser

def normalize_text(text):
    text = text.lower()
    text = re.sub(r'\s+', ' ', text).strip()
    return text

runnablee = RunnableLambda(normalize_text)
chain = runnablee | llm | StrOutputParser() # StrOutputParser == .content
print(chain.invoke("I        AM             OMAR Khattab"))
```

```python
from langchain_core.runnables import RunnableLambda, RunnableParallel

text = 'building large language models with prompt engineering'
title_case = RunnableLambda(lambda text: text.title())
count_words = RunnableLambda(lambda text: len(text.split()))

serial_chain = title_case | count_words
print(serial_chain.invoke(text))  # not meaningful here

parallel_chain = RunnableParallel({
    'title': title_case,
    'word_count': count_words
})
print(parallel_chain.invoke(text))
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel, RunnableLambda

statements = [
    "I had a fantastic time hiking up the mountain yesterday.",
    "I am feeling quite stressed about the upcoming project deadline."
]

sentiment_prompt = ChatPromptTemplate.from_template(
    "Analysis the overall sentiment (Positive,Negative) of the following statement. only output the result:\n\n{statement}"
)
topic_prompt = ChatPromptTemplate.from_template(
    "Identify the main topic of the following statement in a few words. only output the result:\n\n{statement}"
)
followup_prompt = ChatPromptTemplate.from_template(
    "Generate a thoughtful follow-up question related to the following statement. only output the result:\n\n{statement}"
)

sentiment_chain = sentiment_prompt | llm | StrOutputParser()
topic_chain = topic_prompt | llm | StrOutputParser()
followup_chain = followup_prompt | llm | StrOutputParser()

parallel = RunnableParallel({
    "sentiment": sentiment_chain,
    "topic": topic_chain,
    "followup": followup_chain
})


prompts = [parallel.invoke({"statement": s}) for s in statements]
print(prompts[0])
print(prompts[1])

```

```python
# more profesional
inputs = [{"statement": s} for s in statements]
results = parallel.batch(inputs)

for r in results:
    print(r)
```

```python
# Chatbot
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
messages = [SystemMessage(
    content="You are a helpful and friendly assistant for collage professor.")]

while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        print("Goodbye!")
        break
    
    messages.append(HumanMessage(content=user_input))
    response = llm.invoke(messages)
    print("Bot:", response.content)
    messages.append(AIMessage(content=response.content))
```

### Agent&langGraph


```python
from typing_extensions import TypedDict
from typing import Optional, Dict, Any
from langgraph.graph import StateGraph, START, END


# mail scheme
class AcademicEmailState(TypedDict):
    student_mail: str
    grade: float
    is_pass: Optional[bool]
    mail_suggest: Optional[str]
    mail_change: Optional[str]
    send_approval: Optional[bool]
    sent_status: Optional[bool]
    final_message: Optional[str]

# functions
def read_grade(state: AcademicEmailState) -> Dict[str, Any]:
    print(f"[read_grade] Student: {state.get('student_mail')}, Grade: {state.get('grade')}")
    return {}

def classify_grade_node(state: AcademicEmailState) -> Dict[str, Any]:
    grade = state.get("grade", 0.0)
    return {"is_pass": grade >= 50}

def suggest_mail_node(state: AcademicEmailState) -> Dict[str, Any]:
    is_passed = state.get("is_pass", False)
    grade = state.get("grade", 0.0)
    if is_passed:
        mail_suggest = f"Congratulations, you passed! Grade: {grade}"
    else:
        mail_suggest = f"Unfortunately, you did not pass this time. Grade: {grade}"
    return {"mail_suggest": mail_suggest}

def take_changes_node(state: AcademicEmailState) -> Dict[str, Any]:
    suggested = state.get("mail_suggest", "")
    changed = state.get("mail_change")
    final = changed if (changed is not None and changed != "") else suggested
    return {"final_message": final}

def send_mail_node(state: AcademicEmailState) -> Dict[str, Any]:
    print(f"Sending to {state.get('student_mail')}:\n{state.get('final_message')}")
    return {"sent_status": True}

email_graph = StateGraph(AcademicEmailState)

email_graph.add_node("read_grade", read_grade)
email_graph.add_node("classify_grade_node", classify_grade_node)
email_graph.add_node("suggest_mail_node", suggest_mail_node)
email_graph.add_node("take_changes_node", take_changes_node)
email_graph.add_node("send_mail_node", send_mail_node)

email_graph.add_edge(START, "read_grade")
email_graph.add_edge("read_grade", "classify_grade_node")
email_graph.add_edge("classify_grade_node", "suggest_mail_node")
email_graph.add_edge("suggest_mail_node", "take_changes_node")
email_graph.add_edge("take_changes_node", "send_mail_node")
email_graph.add_edge("send_mail_node", END)

compiled_graph = email_graph.compile()
compiled_graph.invoke({
    "student_mail": "omar@student.uni.edu",
    "grade": 85.0,
    "is_pass": None,
    "mail_suggest": None,
    "mail_change": None,
    "send_approval": None,
    "sent_status": None,
    "final_message": None
})
```
